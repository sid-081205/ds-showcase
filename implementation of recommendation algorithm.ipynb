# 🎵 How the Recommendation Algorithm Works

This notebook explains our simple 2-step process for recommending songs that two people will both like.

### Step 1: Combine the Moods
First, we take the music taste of Person A and Person B and find the middle ground (the average).

```python
person_a_mood = [0.8, 0.9, 0.1, 0.2] # 4 traits
person_b_mood = [0.2, 0.1, 0.9, 0.8] # 4 traits

joint_mood = [
    (0.8+0.2)/2, 
    (0.9+0.1)/2, 
    (0.1+0.9)/2, 
    (0.2+0.8)/2
]
print("Our Target Mood:", joint_mood)
```

### Step 2: Find the Closest Match (KNN)
We use **KNN** (K-Nearest Neighbors) to look through our dataset and find songs that are mathematically closest to our **Target Mood** across all features.

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np

# Song dataset (each row is a song with its 4 mood traits)
songs = np.array([
    [0.9, 0.9, 0.1, 0.1],
    [0.1, 0.1, 0.9, 0.9],
    [0.5, 0.5, 0.5, 0.5] # This is the closest to our 0.5 target!
])

model = NearestNeighbors(n_neighbors=1)
model.fit(songs)

distance, index = model.kneighbors([joint_mood])
print("Recommended Song Index:", index)
```
